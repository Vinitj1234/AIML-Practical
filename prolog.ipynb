# **Theory**

### 1. **Introduction to Prolog**

Prolog stands for PROgramming in LOGic. It is a logic programming language widely used for artificial intelligence and knowledge-based systems.

Unlike conventional programming languages, where the programmer specifies a sequence of instructions, Prolog represents knowledge using facts and rules. The user then provides queries, and Prolog uses logical inference, unification, and backtracking to find solutions.

A Prolog program mainly consists of:

* Facts
* Rules
* Queries

For example:

book(java).

is a fact stating that Java is a book.

A rule such as:

borrowed_by(Book,Member) :-
    borrowed(Member,Book).

states that a book is borrowed by a member if the member has borrowed that book.

### 2. **Knowledge Representation**

The system represents information about:

* Books
* Members
* Authors
* Book authors
* Borrowed books

For example:

book(java).
member(vinit).
author(james).

These facts represent that Java is a book, Vinit is a library member, and James is an author.

The relationship between a book and its author is represented using:

written_by(java,james).

The relationship between a member and a borrowed book is represented using:

borrowed(vinit,java).

### 3. **Borrowed Book Checking**

The system uses the borrowed_by/2 predicate to determine which book is borrowed by a particular member.

The rule is:

borrowed_by(Book,Member) :-
    borrowed(Member,Book).

This rule reverses the relationship represented by borrowed/2 and allows the system to find the book from the member.

### 4. **Author Identification**

The author_of/2 predicate determines the author of a particular book.

The rule is:

author_of(Author,Book) :-
    written_by(Book,Author).

This rule checks the written_by/2 relationship and identifies the author of the specified book.

### 5. **Same Author Checking**

The same_author/2 predicate checks whether two books have the same author.

The rule is:

same_author(Book1,Book2) :-
    written_by(Book1,Author),
    written_by(Book2,Author).

The two conditions are connected using logical AND. Therefore, both books must have the same author for the rule to become true.
