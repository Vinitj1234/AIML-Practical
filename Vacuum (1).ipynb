### 1. **Introduction to Intelligent Agents**

An intelligent agent is a system that observes its environment and performs actions according to the current situation.

In this experiment, the Vacuum Cleaner Agent observes the room and performs actions such as moving and cleaning.

The agent maintains information about:

- Battery level
- Current position
- Amount of dirt cleaned

The vacuum cleaner takes decisions based on the condition of the room and its battery.


### 2. **Vacuum Cleaner Agent**

A Vacuum Cleaner Agent is a simple example of an intelligent agent in Artificial Intelligence.

The room is represented using a list. Different symbols represent different conditions:

- `V` represents the initial vacuum position.
- `D` represents dirt.
- `C` represents a clean position.
- `X` represents an obstacle.

For example:

```python
room = ['V', 'D', 'C', 'D', 'D', 'D']

````python
class vacuumCleaner:
    def __init__(self):
        self.battery = 100
        self.position = 0
        self.dirt_cleaned = 0

    def clean(self, room):
        if self.battery > 0:
            if room[self.position] == 'D':
                print("Cleaning in progress...")
                room[self.position] = 'C'
                self.dirt_cleaned += 1
                self.battery -= 5
            else:
                print("No dirt at current position.")
        else:
            print("Battery is empty")

    def move_to(self, target_position, room):
        while self.position < target_position:
            next_position = self.position + 1

            if room[next_position] == 'X':
                print("Obstacle detected! Cannot move forward.")
                return

            if self.battery < 2:
                print("Battery is low!")
                return

            self.position += 1
            self.battery -= 2

            print("Moving to position:", self.position)

        print("Reached target position.")


room = ['V', 'D', 'C', 'D', 'D', 'D']

vacuum = vacuumCleaner()

for position, place in enumerate(room):

    if place == 'D':

        print("Dirt found at position:", position)

        vacuum.move_to(position, room)

        if vacuum.position == position:
            vacuum.clean(room)


print("Cleaning Report")

print("Room:", room)
print("Vacuum Position:", vacuum.position)
print("Battery Remaining:", vacuum.battery)
print("Dirt Cleaned:", vacuum.dirt_cleaned)
````

# **Code Explanation**

### `class vacuumCleaner:`

### `__init__()`

### `clean(room)`

This method cleans the current position if it contains dirt.

When dirt is found:

- `D` is changed to `C`
- Dirt counter increases by 1
- Battery decreases by 5

### `move_to(target_position, room)`

This method moves the vacuum cleaner toward the target position.

For every movement:

- The position increases by 1.
- Battery decreases by 2.
- The program checks for obstacles.
- The program checks the battery level.

### `room`

Here:

### `enumerate(room)`

The `enumerate()` function provides both the position number and the value stored at that position.
