# **Problem Statement**

Find a path from a start position to a goal position in a grid containing obstacles.

BFS, DFS, Greedy Best-First Search, A*, Hill Climbing, and Simulated Annealing are used for pathfinding.

The N-Queens problem is solved using Backtracking.

```python
from collections import deque
import heapq
import time
import random
import math

grid = [
    ['S', 0, 0, 0, 'X'],
    ['X', 0, 'X', 0, 'X'],
    [0, 0, 0, 0, 0],
    [0, 'X', 'X', 'X', 0],
    [0, 0, 0, 0, 'G']
]

start = (0, 0)
goal = (4, 4)

def neighbours(p):
    r, c = p
    moves = [(1,0),(-1,0),(0,1),(0,-1)]
    result = []

    for x, y in moves:
        a, b = r+x, c+y
        if 0 <= a < 5 and 0 <= b < 5 and grid[a][b] != 'X':
            result.append((a,b))

    return result

def h(p):
    return abs(p[0]-goal[0]) + abs(p[1]-goal[1])

print("Grid:")
for row in grid:
    print(row)
```

```python
def bfs():
    q = [(start, [start])]
    visited = {start}
    count = 0

    while q:
        node, path = q.pop(0)
        count += 1

        if node == goal:
            return path, count

        for n in neighbours(node):
            if n not in visited:
                visited.add(n)
                q.append((n, path + [n]))

def dfs():
    stack = [(start, [start])]
    visited = {start}
    count = 0

    while stack:
        node, path = stack.pop()
        count += 1

        if node == goal:
            return path, count

        for n in neighbours(node):
            if n not in visited:
                visited.add(n)
                stack.append((n, path + [n]))

t = time.perf_counter()
bfs_path, bfs_nodes = bfs()
bfs_time = time.perf_counter() - t

t = time.perf_counter()
dfs_path, dfs_nodes = dfs()
dfs_time = time.perf_counter() - t

print("BFS:", bfs_path)
print("Path Length:", len(bfs_path)-1)
print("Nodes:", bfs_nodes)
print("Time:", bfs_time)

print("\nDFS:", dfs_path)
print("Path Length:", len(dfs_path)-1)
print("Nodes:", dfs_nodes)
print("Time:", dfs_time)
```

```python
def greedy():
    q = [(h(start), start, [start])]
    visited = {start}
    count = 0

    while q:
        _, node, path = heapq.heappop(q)
        count += 1

        if node == goal:
            return path, count

        for n in neighbours(node):
            if n not in visited:
                visited.add(n)
                heapq.heappush(q, (h(n), n, path + [n]))

def astar():
    q = [(h(start), 0, start, [start])]
    visited = set()
    count = 0

    while q:
        _, cost, node, path = heapq.heappop(q)
        count += 1

        if node == goal:
            return path, count

        if node in visited:
            continue

        visited.add(node)

        for n in neighbours(node):
            if n not in visited:
                new_cost = cost + 1
                heapq.heappush(q, (new_cost + h(n), new_cost, n, path + [n]))

t = time.perf_counter()
greedy_path, greedy_nodes = greedy()
greedy_time = time.perf_counter() - t

t = time.perf_counter()
astar_path, astar_nodes = astar()
astar_time = time.perf_counter() - t

print("Greedy:", greedy_path)
print("Path Length:", len(greedy_path)-1)
print("Nodes:", greedy_nodes)
print("Time:", greedy_time)

print("\nA*:", astar_path)
print("Path Length:", len(astar_path)-1)
print("Nodes:", astar_nodes)
print("Time:", astar_time)
```

```python
def hill_climbing():
    node = start
    path = [node]
    count = 1

    while node != goal:
        ns = neighbours(node)
        next_node = min(ns, key=h)

        if h(next_node) >= h(node):
            return None, count

        node = next_node
        path.append(node)
        count += 1

    return path, count


def simulated_annealing():
    node = start
    path = [node]
    count = 1
    temp = 10

    while node != goal and count < 100:
        ns = neighbours(node)
        next_node = random.choice(ns)

        if h(next_node) < h(node):
            node = next_node
        else:
            p = math.exp((h(node) - h(next_node)) / temp)
            if random.random() < p:
                node = next_node

        path.append(node)
        count += 1
        temp = temp * 0.95

    return path, count


t = time.perf_counter()
hill_path, hill_nodes = hill_climbing()
hill_time = time.perf_counter() - t

t = time.perf_counter()
sa_path, sa_nodes = simulated_annealing()
sa_time = time.perf_counter() - t

print("Hill Climbing:", hill_path)
print("Nodes:", hill_nodes)
print("Time:", hill_time)

print("\nSimulated Annealing:", sa_path)
print("Nodes:", sa_nodes)
print("Time:", sa_time)
```

```python
n = 8
board = [-1] * n
count = 0

def safe(row, col):
    for i in range(row):
        if board[i] == col:
            return False
        if abs(board[i] - col) == abs(i - row):
            return False
    return True

def solve(row):
    global count
    count += 1

    if row == n:
        return True

    for col in range(n):
        if safe(row, col):
            board[row] = col

            if solve(row + 1):
                return True

            board[row] = -1

    return False

t = time.perf_counter()
found = solve(0)
queen_time = time.perf_counter() - t

print("Solution Found:", found)
print("Nodes:", count)
print("Time:", queen_time)

for i in range(n):
    row = ['.'] * n
    row[board[i]] = 'Q'
    print(' '.join(row))
```
